```python
a=[1,4,2,6,10]
plt.plot(a,"v:c")
```

```python
#Marker Size
a=[1,4,2,6,10]
plt.plot(a,"v:m",ms=20)
```

```python
#Marker Edge Color
a=[1,4,2,6,10]
plt.plot(a,"v:m",ms=10,mec="c")
```

```python
#Marker Edge Color
a=[1,4,2,6,10]
plt.plot(a,"v:g",ms=10,mec="b")
```

```python
#Marker Face Color
a=[1,4,2,6,10]
plt.plot(a,"v:g",ms=10,mec="b",mfc="r")
```

```python
#Changing line color using c
a=[1,4,2,6,10]
plt.plot(a,marker="X",ms=10,mec="b",mfc="r",c="k")
```

```python
#Using Hexadecimal color for Markers
a=np.array([3,8,1,10])
plt.plot(a,marker="X",ms=15,mec="#42f57e",mfc="#e9f542",c="#f54242")
```

```python
#Using Hexadecimal color for Markers
a=np.array([3,8,1,10])
plt.plot(a,marker="^",ms=15,mec="#42f5ec",mfc="#f542e9",c="#42f5d7")
```

```python
#Named color for markers
a=np.array([3,8,1,10])
plt.plot(a,marker="^",ms=15,mec="mediumblue",mfc="maroon",c="lime")
```

```python
#Line style
a=np.array([3,8,1,10])
plt.plot(a,marker="^",ms=15,mec="mediumblue",mfc="maroon",c="lime",linestyle="dotted")
plt.show()
```

```python
a=np.array([3,8,1,10])
plt.plot(a,marker="X",ms=15,mec="#42f57e",mfc="#e9f542",c="#f54242",linestyle="dashed")
```

```python
a=np.array([3,8,1,10])
plt.plot(a,marker="X",ms=15,mec="#42f57e",mfc="#e9f542",c="#f54242",linestyle="--")
```

```python
#line Width
a=np.array([3,8,1,10])
plt.plot(a,marker="^",ms=15,mec="mediumblue",mfc="maroon",c="lime",linewidth=7)
```
