```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
#SIMPLE PIE CHART
x=np.array([35,25,25,15])
plt.pie(x)
plt.title("PIE CHART")
```

```python
#PIE CHART WITH LABELS
x=np.array([35,25,25,15])
my_labels=["Apples","Bananas","Grapes","Mangos"]
plt.pie(x,labels=my_labels,startangle=90)
plt.title("PIE CHART")
```

```python
#Exploded 
x=np.array([35,25,25,15])
my_labels=["Apples","Bananas","Grapes","Mangos"]
myExplode=[0.2,0.1,0,0]
plt.pie(x,labels=my_labels,startangle=90,explode=myExplode)
plt.title("PIE CHART")
```

```python
x=np.array([35,25,25,15])
my_labels=["Apples","Bananas","Grapes","Mangos"]
myExplode=[0.2,0.1,0.1,0.2]
plt.pie(x,labels=my_labels,startangle=90,explode=myExplode)
plt.title("PIE CHART")
```

```python
x=np.array([35,25,25,15])
my_labels=["Apples","Bananas","Grapes","Mangos"]
myExplode=[0.2,0,0,0]
plt.pie(x,labels=my_labels,startangle=90,explode=myExplode)
plt.title("PIE CHART")
```

```python
#SHADOW
x=np.array([35,25,25,15])
my_labels=["Apples","Bananas","Grapes","Mangos"]
my_colors=['red','hotpink','black','yellow']
myExplode=[0.2,0,0,0]
plt.pie(x,labels=my_labels,startangle=90,explode=myExplode,shadow=True,colors=my_colors)
plt.title("PIE CHART")
```

```python
 #PIE CHART WITH LEGEND
x=np.array([35,25,25,15])
my_labels=["Apples","Bananas","Grapes","Mangos"]
plt.pie(x,labels=my_labels)
plt.legend(loc="upper right")
plt.title("PIE CHART")
```

```python
x=np.array([35,25,25,15])
my_labels=["Apples","Bananas","Grapes","Mangos"]
plt.pie(x,labels=my_labels)
plt.legend()
plt.title("PIE CHART")
```
